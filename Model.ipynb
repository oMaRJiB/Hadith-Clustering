# Model Testing

```python
#Preprocesser Directives
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
df = pd.read_csv("data/processed/hadith_eda_export.csv")
```

```python
df = df.dropna(subset=["Arabic_Matn","Arabic_Matn_Stripped"])
```

## TF-IDF + K-Means

```python
tfidf = TfidfVectorizer(analyzer='word',
                        ngram_range=(1,1),
                        min_df=3,
                        max_df=.99,
                        max_features=10000)
X = tfidf.fit_transform(df['Arabic_Matn'])
```

### Experiment 1: Regular Arabic

```python
kMeansDict = {}
for k in range(3,200):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X)
    kMeansDict[k]=labels
    
    score = silhouette_score(
            X, kMeansDict[k],
            sample_size=min(5000, X.shape[0]),
            random_state=42
        )

    sizes = pd.Series(kMeansDict[k]).value_counts()
    print(f"k={k:3} | silhouette={score:.4f} | "
        f"smallest cluster={sizes.min()} | "
        f"largest cluster={sizes.max()}")
    print("Inertia:", kmeans.inertia_)
```

### Experiment 2: Stripped Arabic

```python
tfidf = TfidfVectorizer(analyzer='word',
                        ngram_range=(1,1),
                        min_df=3,
                        max_df=.99,
                        max_features=10000)
X = tfidf.fit_transform(df['Arabic_Matn_Stripped'])
kMeansDictStripped = {}
for k in range(3,200):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X)
    
    score = silhouette_score(
            X, labels,
            sample_size=min(5000, X.shape[0]),
            random_state=42
        )

    sizes = pd.Series(labels).value_counts()
    kMeansDictStripped[k]={'labels':labels,
                           'silhouette':score,
                           'smallest':sizes.min(),
                           'largest':sizes.max(),
                           'inertia': kmeans.inertia_}
    print(f"k={k:3} | silhouette={score:.4f} | "
        f"smallest cluster={sizes.min()} | "
        f"largest cluster={sizes.max()}")
    print("Inertia:", kmeans.inertia_)
```
