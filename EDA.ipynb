# Exploratory Data Analysis

```python
#Preprocessor Directives
import pandas as pd
import numpy as np
import glob
import os.path
import re
```

```python
pd.set_option('display.max_colwidth', None)
```

## File Access

```python
#Get File names
PATH ="data/raw/LK-Hadith-Corpus" # Path of the corpus file
book_filenames = sorted(glob.glob(PATH +'//**//*.csv', recursive=True)) # read all csv files in all books 
book_filenames
```

```python
hadiths = []
for book in book_filenames:
    temp = pd.read_csv(book)
    temp['Collection'] =  os.path.basename(os.path.dirname(book))
    hadiths.append(temp)

df: pd.DataFrame = pd.concat(hadiths, ignore_index=True) 
```

## Dataset info

```python
print("Shape:", df.shape)
display(df.dtypes)

display(df["Collection"].value_counts())

display(df.columns.tolist())
```

```python
df.describe(include='all')
```

```python
df.head()
```

```python
#Quasi Constant Columns
threshold = 0.99

for col in df.columns:
    most_frequent_ratio = df[col].value_counts(normalize=True, dropna=False).iloc[0]

    if(most_frequent_ratio >= threshold):
        print(f"{col}: {most_frequent_ratio:.2%}; Unique values:{df[col].unique().tolist()}")
    if(most_frequent_ratio==1):
        df.drop([col,],axis=1,inplace=True)
        print("DROPPED")
```

## Checking for duplicate hadiths

```python
print("Duplicate rows:", df.duplicated().sum())

print(
    "Duplicate collection/Hadith number pairs:",
    df.duplicated(["Collection", "Hadith_number"]).sum()
)
```

```python
#Investigating duplicate hadiths
duplicate_ids = df[
    df.duplicated(
        subset=["Collection","Chapter_Number","Section_Number", "Hadith_number"],
        keep=False
    )
].sort_values(["Collection","Chapter_Number","Section_Number", "Hadith_number"])

display(
    duplicate_ids[
        [
            "Collection",
            "Chapter_Number",
            "Section_Number",
            "Hadith_number",
            "Chapter_English",
            "Arabic_Hadith",
            "Arabic_Matn"
        ]
    ]
)
```

```python
# Count repeated identifier groups and affected records
duplicate_summary = (
    duplicate_ids
    .groupby(["Collection","Chapter_Number","Section_Number", "Hadith_number"])
    .agg(
        record_count=("Hadith_number", "size"),
        chapter_count=("Chapter_Number", "nunique"),
        distinct_arabic_hadith=("Arabic_Hadith", "nunique"),
        distinct_arabic_matn=("Arabic_Matn", "nunique")
    )
    .reset_index()
)

print("Repeated identifier groups:", len(duplicate_summary))
print("Records involved:", len(duplicate_ids))
print("\nSummary:")
display(duplicate_summary)
```

```python
missing_matn = df[df["Arabic_Matn"].isna()]

same_text = (
    missing_matn["Arabic_Hadith"].str.strip()
    == missing_matn["Arabic_Isnad"].str.strip()
)

print("Missing matn records:", len(missing_matn))
print("Hadith identical to isnad:", same_text.sum())
print("Hadith differs from isnad:", (~same_text).sum())
print("Isnad also missing:", missing_matn["Arabic_Isnad"].isna().sum())
```

```python
#Filling missing matns with full hadiths
df["Arabic_Matn"] = df["Arabic_Matn"].fillna(df["Arabic_Hadith"])
```

```python
#Get unique chapter names
keywords = ['Adhan','Cloth','Divorce','Prayer','Faith','Food','Vows','Marriage','Wills']
uniqueChapters = df["Chapter_English"].unique()

uniqueChapters = np.unique([
    next(
        (keyword for keyword in keywords
         if keyword.lower() in chapter.lower()),
        chapter
    )
    for chapter in uniqueChapters
])

uniqueChapters = np.array([x.replace("the ", "").replace("The ", "").replace("'",'') for x in uniqueChapters])
uniqueChapters = np.vectorize(lambda x: re.sub(r"(?:Book|Chapters?|Description)\s+(?:(?:of|on|regarding|concerning)\s+)?|\s*\([^)]*\)|\b(?:A[a-z]-|ال)", "", str(x),flags=re.I).strip())(uniqueChapters)

uniqueChapters=np.unique(uniqueChapters)
print("Unique chapters:", len(uniqueChapters))
uniqueChapters
```
